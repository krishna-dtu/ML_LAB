```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("Titanic-Dataset.csv")
df.head()
```

```python
print("Dataset Shape: ")
print(df.shape)
print("\nColumn Names: ")
print(df.columns.tolist())
```

```python
print("\nDataset Information: ")
df.info
```

```python
print("\nStatistical Information: ")
print(df.describe(include='all'))
```

```python
print("Duplicated Rows: ", df.duplicated().sum())
```

```python
print("Missing Values: \n", df.isnull().sum())
```

```python
print("Categorical Columns :\n",df.select_dtypes(include=['object']).columns.tolist())
```

```python
print("Unique Values: \n")
for col in df.columns:
  if df[col].nunique()==1:
    print(col)
```

```python
X = df.drop("Survived", axis=1)
y = df["Survived"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.30, random_state=42)
print('Training Set: ', X_train.shape)
print("Testing Set: ", X_test.shape)
```

```python
dummy_mf=DummyClassifier(strategy='most_frequent')
dummy_mf.fit(X_train, y_train)
pred_mf=dummy_mf.predict(X_test)
acc_mf=accuracy_score(y_test, pred_mf)
print("Most frequent baseline accuracy : ",acc_mf)
```

```python
dummy_str=DummyClassifier(strategy='stratified', random_state=42)
dummy_str.fit(X_train, y_train)
pred_str=dummy_str.predict(X_test)
acc_str=accuracy_score(y_test, pred_str)
print("Stratified accuracy : ", acc_str)
```

```python
results=pd.DataFrame({"Strategy":["Most Frequent", "Stratified"], "Accuracy":[acc_mf, acc_str]})
print(results)
```

## Logistic Regression Model

Unlike the baseline models, Logistic Regression learns patterns from the input features and attempts to predict passenger survival based on the available information.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
X_num=df.select_dtypes(include=['int64', 'float64'])
X_num=X_num.drop("Survived",axis=1)
imputer=SimpleImputer(strategy='mean')
X_num=imputer.fit_transform(X_num)
y=df["Survived"]
```

```python
X_train_lr, X_test_lr, y_train_lr, y_test_lr=train_test_split(X_num,y, test_size=0.30, random_state=42)
lr_model=LogisticRegression(max_iter=1000)
lr_model.fit(X_train_lr,y_train_lr)
lr_pred=lr_model.predict(X_test_lr)
lr_acc=accuracy_score(y_test_lr, lr_pred)
print("Logistic Regression Accuracy: ", lr_acc)
```

```python
results=pd.DataFrame({'Model':["Dummy (most frequent)", "Dummy(stratified)", "Logistic Regression"], 'Accuracy':[acc_mf,acc_str,lr_acc]})
print(results)
```

## Model Comparison

The DummyClassifier models provide baseline performance by making simple predictions without learning from the data.

The Logistic Regression model learns relationships between passenger features and survival outcomes. Its accuracy is significantly higher than the baseline models, indicating that the dataset contains useful predictive information.

Since Logistic Regression outperforms both dummy strategies, applying a real machine learning model is justified for this problem.
